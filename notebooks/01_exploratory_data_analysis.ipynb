```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
project_root = Path.cwd()

df = pd.read_csv(
    project_root / ".." / "data" / "raw" /
    "default_credit_clients_raw.csv"
)

print("Dataset carregado!")
print(f"Dimensões: {df.shape}")

df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": (
        df.isnull().mean() * 100
    )
})

missing.sort_values(
    by="missing_percent",
    ascending=False
)
```

```python
print(
    "Registros duplicados:",
    df.duplicated().sum()
)
```

```python
print(df.columns.tolist())
```

```python
target = "Y"

print(df[target].value_counts())

print("\nProporção percentual:")
print(
    df[target].value_counts(normalize=True) * 100
)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x=target
)

plt.title("Distribuição da variável-alvo")
plt.xlabel("Inadimplência no próximo mês")
plt.ylabel("Quantidade de clientes")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="X5",
    bins=30,
    kde=True
)

plt.title("Distribuição de idade dos clientes")
plt.xlabel("Idade")
plt.ylabel("Frequência")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.boxplot(
    data=df,
    x="Y",
    y="X5"
)

plt.title("Idade por situação de inadimplência")
plt.xlabel("Inadimplência")
plt.ylabel("Idade")

plt.show()
```

```python
categorical_columns = [
    "X2",  # Sexo
    "X3",  # Educação
    "X4"   # Estado civil
]

for column in categorical_columns:

    print(f"\nDistribuição: {column}")

    print(
        df[column].value_counts().sort_index()
    )
```

```python
column_mapping = {
    "X1": "LIMIT_BAL",
    "X2": "SEX",
    "X3": "EDUCATION",
    "X4": "MARRIAGE",
    "X5": "AGE",
    "X6": "PAY_0",
    "X7": "PAY_2",
    "X8": "PAY_3",
    "X9": "PAY_4",
    "X10": "PAY_5",
    "X11": "PAY_6",
    "X12": "BILL_AMT1",
    "X13": "BILL_AMT2",
    "X14": "BILL_AMT3",
    "X15": "BILL_AMT4",
    "X16": "BILL_AMT5",
    "X17": "BILL_AMT6",
    "X18": "PAY_AMT1",
    "X19": "PAY_AMT2",
    "X20": "PAY_AMT3",
    "X21": "PAY_AMT4",
    "X22": "PAY_AMT5",
    "X23": "PAY_AMT6",
    "Y": "DEFAULT_NEXT_MONTH"
}

df_eda = df.rename(columns=column_mapping).copy()

df_eda.head()
```

```python
print("Duplicatas completas:")
print(df_eda.duplicated().sum())
```

```python
categorical_columns = [
    "SEX",
    "EDUCATION",
    "MARRIAGE"
]

for column in categorical_columns:
    print(f"\n{column}")
    print(df_eda[column].value_counts().sort_index())
```

```python
numerical_columns = [
    "LIMIT_BAL",
    "AGE",
    "BILL_AMT1",
    "BILL_AMT2",
    "PAY_AMT1",
    "PAY_AMT2"
]

df_eda[numerical_columns].describe().T
```

```python
# Visualizar os registros duplicados
duplicates = df_eda[
    df_eda.duplicated(keep=False)
].sort_values(by="AGE")

print(f"Registros envolvidos: {len(duplicates)}")

display(duplicates.head(20))
```

```python
duplicate_analysis = (
    df_eda.groupby(list(df_eda.columns))
    .size()
    .reset_index(name="frequency")
)

duplicate_analysis = duplicate_analysis[
    duplicate_analysis["frequency"] > 1
]

duplicate_analysis[
    ["DEFAULT_NEXT_MONTH", "frequency"]
]
```
