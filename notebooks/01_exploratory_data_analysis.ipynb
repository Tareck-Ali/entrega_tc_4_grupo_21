```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

sns.set_theme(style="whitegrid")

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
```

```python
project_root = Path.cwd()

df = pd.read_csv(
    project_root / ".." / "data" / "raw" /
    "default_credit_clients_raw.csv"
)

print("Dataset carregado!")
print(f"Dimensões: {df.shape}")

df.head()
```

```python
df.info()
```

```python
df.describe().T
```

```python
missing = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percent": (
        df.isnull().mean() * 100
    )
})

missing.sort_values(
    by="missing_percent",
    ascending=False
)
```

```python
print(
    "Registros duplicados:",
    df.duplicated().sum()
)
```

```python
print(df.columns.tolist())
```

```python
target = "Y"

print(df[target].value_counts())

print("\nProporção percentual:")
print(
    df[target].value_counts(normalize=True) * 100
)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x=target
)

plt.title("Distribuição da variável-alvo")
plt.xlabel("Inadimplência no próximo mês")
plt.ylabel("Quantidade de clientes")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(
    data=df,
    x="X5",
    bins=30,
    kde=True
)

plt.title("Distribuição de idade dos clientes")
plt.xlabel("Idade")
plt.ylabel("Frequência")

plt.show()
```

```python
plt.figure(figsize=(12, 5))

sns.boxplot(
    data=df,
    x="Y",
    y="X5"
)

plt.title("Idade por situação de inadimplência")
plt.xlabel("Inadimplência")
plt.ylabel("Idade")

plt.show()
```

```python
categorical_columns = [
    "X2",  # Sexo
    "X3",  # Educação
    "X4"   # Estado civil
]

for column in categorical_columns:

    print(f"\nDistribuição: {column}")

    print(
        df[column].value_counts().sort_index()
    )
```
